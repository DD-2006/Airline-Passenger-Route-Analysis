```python
#Classify flights based on delay duration, load factor, and passenger satisfaction using conditional logic to identify low-, medium-, and high-performing flights.
```

```python
load_category = []

for load in flights1["Load_Factor_Pct"]:
    if load <= 50:
        load_category.append("Low Load")
    elif load <= 80:
        load_category.append("Medium Load")
    else:
        load_category.append("High Load")

flights1["Load_Category"] = load_category
```

```python
flights1["Load_Category"].head(10)
```

```python
#Analyze flight records to identify and count flights experiencing significant delays, particularly flights delayed by more than 30 minutes.
```

```python
count = 0

for _, flight in flights1.iterrows():
    if flight["Delay_Minutes"] > 30:
        count += 1
        print(
            "Flight ID:", flight["Flight_ID"],
            "| Delay:", flight["Delay_Minutes"], "minutes"
        )

print("Total flights delayed by more than 30 minutes:", count)
```

```python
#Use a for loop to iterate through staff records and identify and count employees whose salary exceeds a specified threshold.
```

```python
count = 0

for _, staff in staffs1.iterrows():
    if staff["Salary_USD"] > 80000:
        count += 1

print("Number of staff earning above $80,000:", count)
```

```python
#Use a while loop to process staff records and identify the first few employees with experience above a specified number of years.
```

```python
count = 0
i = 0
required_staff = 5
experience_limit = 15

while count < required_staff and i < len(staffs1):

    if staffs1.iloc[i]["Experience_Years"] > experience_limit:
        print(
            "Staff ID:", staffs1.iloc[i]["Staff_ID"],
            "| Name:", staffs1.iloc[i]["Staff_Name"],
            "| Experience:", staffs1.iloc[i]["Experience_Years"], "years"
        )
        count += 1

    i += 1
```

```python
#Use Pandas groupby() to calculate total revenue, passenger count, average load factor, and average delay for each route.
```

```python
route_analysis = flights1.groupby(["Origin_Airport_ID", "Destination_Airport_ID"]).agg(
    Total_Revenue=("Revenue_USD", "sum"),
    Passenger_Count=("Passengers", "sum"),
    Avg_Load_Factor=("Load_Factor_Pct", "mean"),
    Avg_Delay=("Delay_Minutes", "mean")).reset_index()

print(route_analysis.head())
```

```python
#Merge Flights, Passengers, Routes, Airlines, and Airports datasets to create an integrated dataset for cross-dimensional analysis of 
#routes, airlines, delays, and passenger satisfaction.
```

```python
integrated = flights1.merge(
    airlines1[["Airline_ID", "Airline_Name"]],
    on="Airline_ID",
    how="left"
)
```

```python
integrated = integrated.merge(
    passengers1[
        ["Passenger_ID", "Flight_ID", "Travel_Class",
         "Ticket_Price_USD", "Satisfaction_Score", "Satisfaction_Level"]
    ],
    on="Flight_ID",
    how="left"
)
```

```python
integrated = integrated.merge(
    routes1[
        ["Route_ID", "Origin_Airport_ID", "Destination_Airport_ID",
         "Total_Revenue_USD", "Avg_Load_Factor_Pct", "Avg_Delay_Minutes"]
    ],
    on=["Origin_Airport_ID", "Destination_Airport_ID"],
    how="left"
)
```

```python
integrated = integrated.merge(
    airports1[["Airport_ID", "City", "IATA_Code"]],
    left_on="Origin_Airport_ID",
    right_on="Airport_ID",
    how="left"
).rename(columns={
    "City": "Origin_City",
    "IATA_Code": "Origin_IATA"
}).drop(columns="Airport_ID")
```

```python
integrated = integrated.merge(
    airports1[["Airport_ID", "City", "IATA_Code"]],
    left_on="Destination_Airport_ID",
    right_on="Airport_ID",
    how="left"
).rename(columns={
    "City": "Destination_City",
    "IATA_Code": "Destination_IATA"
}).drop(columns="Airport_ID")
```

```python
integrated["Route"] = (
    integrated["Origin_City"] + " → " +
    integrated["Destination_City"]
)
```

```python
#IQR analysis
```

```python
Q1 = flights1["Revenue_USD"].quantile(0.25)
Q3 = flights1["Revenue_USD"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Revenue IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
revenue_outliers = flights1[
    (flights1["Revenue_USD"] < lower_limit) |
    (flights1["Revenue_USD"] > upper_limit)
]

print("Revenue Outliers:", len(revenue_outliers))
```

```python
#Correlation analysis
```

```python
correlation = flights1[
    [
        "Revenue_USD",
        "Load_Factor_Pct",
        "Flight_Duration_Min",
        "Base_Fare_USD",
        "Delay_Minutes"
    ]
].corr()

print(correlation)
```

```python
flight_satisfaction = passengers1.groupby(
    "Flight_ID"
)["Satisfaction_Score"].mean().reset_index()

flight_satisfaction.rename(
    columns={"Satisfaction_Score": "Avg_Satisfaction"},
    inplace=True
)
```

```python
analysis_df = flights1.merge(
    flight_satisfaction,
    on="Flight_ID",
    how="left"
)
```

```python
correlation = analysis_df[
    [
        "Revenue_USD",
        "Load_Factor_Pct",
        "Flight_Duration_Min",
        "Base_Fare_USD",
        "Delay_Minutes",
        "Avg_Satisfaction"
    ]
].corr()

print(correlation)
```

```python
#Correlation heatmap
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation of Airline Performance Metrics")
plt.show()
```
